```python
features = [
    'Country Code',
    'City',
    'Longitude',
    'Latitude',
    'Cuisines',
    'Average Cost for two',
    'Currency',
    'Has Table booking',
    'Has Online delivery',
    'Is delivering now',
    'Price range',
    'Votes'
]

X = df[features]
y = df['Aggregate rating']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
print("Categorical columns:")
print(X.select_dtypes(include='object').columns.tolist())

print("\nNumerical columns:")
print(X.select_dtypes(exclude='object').columns.tolist())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

print("Scikit-learn imported successfully!")
```

```python
categorical_features = X.select_dtypes(include='object').columns.tolist()
numerical_features = X.select_dtypes(exclude='object').columns.tolist()

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('categorical', OneHotEncoder(handle_unknown='ignore'), categorical_features),
        ('numerical', 'passthrough', numerical_features)
    ]
)

print("Preprocessor created successfully!")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.linear_model import LinearRegression

print("Linear Regression imported successfully!")
```

```python
model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('regressor', LinearRegression())
    ]
)

print("Model pipeline created successfully!")
```

```python
model.fit(X_train, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test)

print("Predictions generated!")
```

```python
print(y_pred[:10])
```

```python
comparison = pd.DataFrame({
    'Actual Rating': y_test.values[:10],
    'Predicted Rating': y_pred[:10]
})

print(comparison)
```

```python
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('categorical',
         OneHotEncoder(handle_unknown='ignore'),
         categorical_features),

        ('numerical',
         'passthrough',
         numerical_features)
    ]
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
```

```python
model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('regressor', LinearRegression())
    ]
)
```

```python
tree = tree_model.named_steps['regressor']
```

```python
encoder = tree_model.named_steps['preprocessor']

feature_names = encoder.get_feature_names_out()

print("Number of features:", len(feature_names))
```

```python
importance = tree.feature_importances_

feature_importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance
})

feature_importance_df = feature_importance_df.sort_values(
    by='Importance',
    ascending=False
)

print(feature_importance_df.head(15))
```

```python
top_features = feature_importance_df.head(10)

plt.figure(figsize=(10,6))

plt.barh(
    top_features['Feature'][::-1],
    top_features['Importance'][::-1]
)

plt.xlabel('Importance')
plt.ylabel('Feature')
plt.title('Top 10 Features Influencing Restaurant Ratings')

plt.show()
```

```python
importance_df = feature_importance_df.copy()

importance_df['Original Feature'] = (
    importance_df['Feature']
    .str.replace('numerical__', '', regex=False)
    .str.replace('categorical__', '', regex=False)
    .str.split('_', n=1)
    .str[0]
)

grouped_importance = (
    importance_df
    .groupby('Original Feature')['Importance']
    .sum()
    .sort_values(ascending=False)
)

print(grouped_importance)
```

```python
top_grouped = grouped_importance.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_grouped.index[::-1],
    top_grouped.values[::-1]
)

plt.xlabel('Total Feature Importance')
plt.ylabel('Feature')
plt.title('Top Features Influencing Restaurant Ratings')

plt.show()
```

```python
def predict_restaurant_rating(
    country_code,
    city,
    longitude,
    latitude,
    cuisines,
    average_cost,
    currency,
    table_booking,
    online_delivery,
    delivering_now,
    price_range,
    votes
):

    restaurant = pd.DataFrame([{
        'Country Code': country_code,
        'City': city,
        'Longitude': longitude,
        'Latitude': latitude,
        'Cuisines': cuisines,
        'Average Cost for two': average_cost,
        'Currency': currency,
        'Has Table booking': table_booking,
        'Has Online delivery': online_delivery,
        'Is delivering now': delivering_now,
        'Price range': price_range,
        'Votes': votes
    }])

    prediction = tree_model.predict(restaurant)[0]

    # Keep prediction within the possible rating range
    prediction = np.clip(prediction, 0, 5)

    return round(prediction, 2)
```

```python
import ipywidgets as widgets
from IPython.display import display

# Input fields
country_code = widgets.IntText(
    value=1,
    description='Country Code:'
)

city = widgets.Text(
    value='New Delhi',
    description='City:'
)

longitude = widgets.FloatText(
    value=77.1025,
    description='Longitude:'
)

latitude = widgets.FloatText(
    value=28.7041,
    description='Latitude:'
)

cuisines = widgets.Text(
    value='North Indian',
    description='Cuisines:'
)

average_cost = widgets.IntText(
    value=800,
    description='Avg Cost:'
)

currency = widgets.Text(
    value='Indian Rupees(Rs.)',
    description='Currency:'
)

table_booking = widgets.Dropdown(
    options=['Yes', 'No'],
    value='Yes',
    description='Table Booking:'
)

online_delivery = widgets.Dropdown(
    options=['Yes', 'No'],
    value='Yes',
    description='Online Delivery:'
)

delivering_now = widgets.Dropdown(
    options=['Yes', 'No'],
    value='Yes',
    description='Delivering Now:'
)

price_range = widgets.IntSlider(
    value=3,
    min=1,
    max=4,
    step=1,
    description='Price Range:'
)

votes = widgets.IntText(
    value=500,
    description='Votes:'
)

button = widgets.Button(
    description='Predict Rating'
)

output = widgets.Output()


def predict_button_clicked(b):
    with output:
        output.clear_output()

        rating = predict_restaurant_rating(
            country_code.value,
            city.value,
            longitude.value,
            latitude.value,
            cuisines.value,
            average_cost.value,
            currency.value,
            table_booking.value,
            online_delivery.value,
            delivering_now.value,
            price_range.value,
            votes.value
        )

        print("================================")
        print("     RESTAURANT RATING")
        print("================================")
        print()
        print("Predicted Rating:", rating, "/ 5")


button.on_click(predict_button_clicked)

display(
    country_code,
    city,
    longitude,
    latitude,
    cuisines,
    average_cost,
    currency,
    table_booking,
    online_delivery,
    delivering_now,
    price_range,
    votes,
    button,
    output
)
```

# Cognifyz Task 1 – Predict Restaurant Ratings

## 1. Objective

The objective of this project is to build a machine learning regression model
that predicts the aggregate rating of a restaurant based on other restaurant
features.

## 2. Dataset

The dataset contains information about 9,551 restaurants and 21 columns.

Important features include:

- Restaurant Name
- Country Code
- City
- Longitude
- Latitude
- Cuisines
- Average Cost for two
- Currency
- Table Booking
- Online Delivery
- Price Range
- Votes
- Aggregate Rating

The target variable is:

**Aggregate rating**

## 3. Data Preprocessing

The dataset was checked for missing values.

Only the Cuisines column contained missing values, with 9 missing records.
These values were replaced with "Unknown".

The dataset contained 2,148 restaurants with an aggregate rating of 0.
These restaurants were excluded from the machine learning dataset because
they did not have an actual rating.

After preprocessing:

- Original records: 9,551
- Records used for machine learning: 7,403
- Features used: 12

## 4. Feature Preparation

The dataset contained both numerical and categorical features.

Numerical features included:

- Country Code
- Longitude
- Latitude
- Average Cost for two
- Price Range
- Votes

Categorical features included:

- City
- Cuisines
- Currency
- Has Table booking
- Has Online delivery
- Is delivering now

Categorical features were converted into numerical form using One-Hot Encoding.

## 5. Training and Testing

The dataset was divided into:

- 80% training data
- 20% testing data

The training data was used to train the models, while the testing data was
used to evaluate their performance on previously unseen restaurants.

## 6. Machine Learning Models

Two regression algorithms were tested:

1. Linear Regression
2. Decision Tree Regression

## 7. Model Performance

| Model | MSE | RMSE | R² Score |
|---|---:|---:|---:|
| Linear Regression | 0.1485 | 0.3854 | 0.5197 |
| Decision Tree | 0.1342 | 0.3663 | 0.5661 |

The Decision Tree performed better because it had a lower MSE and RMSE and a
higher R² score.

Therefore, the Decision Tree Regression model was selected as the final model.

## 8. Feature Importance

Feature importance analysis showed that the most influential features were:

1. Votes
2. Longitude
3. Cuisines
4. Latitude
5. Average Cost for two
6. Currency
7. City
8. Price range

The most influential feature was Votes, with an importance of approximately
0.618.

Feature importance indicates how strongly the trained model relied on a
feature when making predictions. It does not necessarily indicate a causal
relationship.

## 9. Prediction System

A prediction function was created using the trained Decision Tree model.

A user can provide restaurant information such as:

- City
- Location
- Cuisine
- Average Cost
- Currency
- Table Booking
- Online Delivery
- Price Range
- Votes

The model then produces a predicted restaurant rating between 0 and 5.

## 10. Conclusion

A machine learning system was successfully developed to predict restaurant
ratings.

Among the tested models, Decision Tree Regression achieved the best
performance with an R² score of approximately 0.5661 and an RMSE of
approximately 0.3663.

The model can be used as a basic restaurant-rating prediction system and can
be further improved using additional features, hyperparameter tuning, and
more advanced machine learning algorithms.
