```python
import pandas as pd

df = pd.read_csv("Dataset .csv")
```

```python
df['Cuisines'] = df['Cuisines'].fillna('Unknown')
```

```python
# Fix the first column name
df.rename(columns={'ï»¿Restaurant ID': 'Restaurant ID'}, inplace=True)

print(df.columns.tolist())
```

```python
def recommend_restaurants(cuisine, price_range, min_rating, city=None):

    recommendations = df[
        (df['Cuisines'].str.contains(cuisine, case=False, na=False)) &
        (df['Price range'] == price_range) &
        (df['Aggregate rating'] >= min_rating)
    ]

    if city:
        recommendations = recommendations[
            recommendations['City'].str.contains(city, case=False, na=False)
        ]

    recommendations = recommendations.sort_values(
        by=['Aggregate rating', 'Votes'],
        ascending=False
    )

    return recommendations[
        ['Restaurant Name', 'City', 'Cuisines',
         'Price range', 'Aggregate rating', 'Votes']
    ].head(10)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
df['Cuisines'] = df['Cuisines'].fillna('')
df['City'] = df['City'].fillna('')

df['Price_range_text'] = df['Price range'].astype(str)

df['features'] = (
    df['Cuisines'] + ' ' +
    df['City'] + ' ' +
    'price_' + df['Price_range_text']
)

df[['Restaurant Name', 'Cuisines', 'City', 'Price range', 'features']].head()
```

```python
tfidf = TfidfVectorizer(stop_words='english')

tfidf_matrix = tfidf.fit_transform(df['features'])

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```

```python
def content_based_recommendation(cuisine, price_range, city=None, min_rating=0):

    # Create user preference profile
    user_profile = cuisine

    if city:
        user_profile += " " + city

    user_profile += " price_" + str(price_range)

    # Convert user profile to TF-IDF
    user_vector = tfidf.transform([user_profile])

    # Calculate similarity
    similarity_scores = cosine_similarity(
        user_vector,
        tfidf_matrix
    ).flatten()

    # Copy dataset
    recommendations = df.copy()

    # Add similarity score
    recommendations['Similarity Score'] = similarity_scores

    # Apply minimum rating
    recommendations = recommendations[
        recommendations['Aggregate rating'] >= min_rating
    ]

    # Sort by similarity and rating
    recommendations = recommendations.sort_values(
        by=['Similarity Score', 'Aggregate rating'],
        ascending=False
    )

    # Return top 10
    return recommendations[
        ['Restaurant Name',
         'City',
         'Cuisines',
         'Price range',
         'Aggregate rating',
         'Votes',
         'Similarity Score']
    ].head(10)
```

```python
content_based_recommendation(
    cuisine="Indian",
    price_range=2,
    city="Hyderabad",
    min_rating=3
)
```

```python
content_based_recommendation(
    cuisine="Italian",
    price_range=2,
    city="Mumbai",
    min_rating=3
)
```

```python
cuisine = input("Enter your preferred cuisine: ")
price_range = int(input("Enter price range (1-4): "))
city = input("Enter your preferred city (or press Enter to skip): ")
min_rating = float(input("Enter minimum rating: "))

results = content_based_recommendation(
    cuisine=cuisine,
    price_range=price_range,
    city=city if city else None,
    min_rating=min_rating
)

print("\n===== RECOMMENDED RESTAURANTS =====")

if results.empty:
    print("No restaurants found matching your preferences.")
else:
    display(results)
```

```python
def evaluate_recommendations(results, cuisine, price_range, min_rating):

    if results.empty:
        print("No recommendations found.")
        return

    cuisine_matches = results['Cuisines'].str.contains(
        cuisine,
        case=False,
        na=False
    ).sum()

    price_matches = (results['Price range'] == price_range).sum()

    rating_matches = (
        results['Aggregate rating'] >= min_rating
    ).sum()

    total = len(results)

    print("===== RECOMMENDATION EVALUATION =====")
    print("Total recommendations:", total)
    print("Cuisine matches:", cuisine_matches)
    print("Price range matches:", price_matches)
    print("Rating matches:", rating_matches)

    print("\nAverage Rating:",
          round(results['Aggregate rating'].mean(), 2))

    print("Average Similarity Score:",
          round(results['Similarity Score'].mean(), 3))
```

```python
evaluate_recommendations(
    results,
    cuisine="Indian",
    price_range=2,
    min_rating=3
)
```

```python
final_results = results[
    [
        'Restaurant Name',
        'City',
        'Cuisines',
        'Price range',
        'Aggregate rating',
        'Votes',
        'Similarity Score'
    ]
]

display(final_results)
```
